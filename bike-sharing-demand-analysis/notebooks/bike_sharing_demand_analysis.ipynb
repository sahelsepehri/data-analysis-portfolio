## 1. Introduction

Bike-sharing services need to understand when people are more likely to rent bikes. Demand can change throughout the year and may be related to weather conditions, working days, and user behavior.

In this project, I analyze daily bike rental data to find patterns in demand and understand how rentals change under different conditions.

**Business Question**

## 2. Dataset Overview

The dataset contains daily bike rental records from 2011 to 2012, with 731 daily observations and 16 columns.

It includes information about the date, season, month, holidays, working days, weather conditions, temperature, humidity, wind speed, and the number of bike rentals.

Some important columns are:

- `dteday`: The date of each observation.
- `season` and `mnth`: The season and month of the year.
- `workingday`: Indicates whether the day is a working day.
- `weathersit`: The weather condition for that day.
- `temp` and `atemp`: Normalized temperature and feeling temperature.
- `hum` and `windspeed`: Humidity and wind speed.
- `casual`: Rentals by casual users.
- `registered`: Rentals by registered users.
- `cnt`: Total bike rentals for the day.

The `cnt` column is the main measure of demand in this analysis.

## 3. Distribution of Daily Bike Rental Demand

- Let's look at how many bikes people rented each day and how much the numbers changed from one day to another.

- **Business question:** Were daily rentals usually around the average, or did they change a lot from day to day?



```python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="cnt",
    bins=20,
    color="#2EC4B6",
    edgecolor="white",
    linewidth=0.8
)

plt.title("Distribution of Daily Bike Rentals", fontsize=14)
plt.xlabel("Number of Rentals")
plt.ylabel("Number of Days")

sns.despine()
plt.tight_layout()
plt.show()
```

## 4. Monthly Bike Rental Demand

Let's look at how bike rentals change from month to month. This helps us understand when demand is usually higher or lower during the year.

**Business question:** Which months have the highest and lowest average daily bike rentals?


```python

# Calculate average daily rentals for each month
monthly_avg = (
    df.groupby("mnth", as_index=False)["cnt"]
    .mean()
)

month_labels = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

plt.figure(figsize=(10, 5))

ax = sns.lineplot(
    data=monthly_avg,
    x="mnth",
    y="cnt",
    marker="o",
    linewidth=2.5,
    color="#2EC4B6"
)

plt.xticks(range(1, 13), month_labels)

plt.title("Average Daily Bike Rentals by Month")
plt.xlabel("Month")
plt.ylabel("Average Daily Rentals")

sns.despine()
plt.tight_layout()
plt.show()
```

- Average daily rentals increased from January to June, stayed high through September, and then declined toward December.
- January had the lowest average, at around 2,176 rentals per day, while June had the highest average, at around 5,772.

## 5. Bike Rentals on Working Days vs. Weekends

- Let's compare daily bike rentals on working days and weekends or holidays.

- **Business question:** Does bike rental demand differ between working days and non-working days?

```python

day_type_avg = (
    df.assign(
        day_type=df["workingday"].map({
            1: "Working day",
            0: "Non-working day"
        })
    )
    .groupby("day_type", as_index=False)["cnt"]
    .mean()
)

day_order = ["Working day", "Non-working day"]

plt.figure(figsize=(8, 5))

ax = sns.barplot(
    data=day_type_avg,
    x="day_type",
    y="cnt",
    hue="day_type",
    order=day_order,
    hue_order=day_order,
    palette={
        "Working day": "#FF6B6B",
        "Non-working day": "#9B5DE5"
    },
    errorbar=None,
    legend=False
)

for container in ax.containers:
    ax.bar_label(container, fmt="%.0f", padding=4)

plt.title("Average Bike Rentals: Working vs Non-working Days")
plt.xlabel("")
plt.ylabel("Average Daily Rentals")

sns.despine()
plt.tight_layout()
plt.show()

```

- Average daily rentals were higher on working days (4,585) than on non-working days (4,330). The difference was around 255 rentals per day.

## 7. Bike Rentals by Weather

- Let's compare bike rentals under different weather conditions.
- This can help us understand whether people tend to rent more bikes when the weather is clear.
- **Business question:** How does average daily bike rental demand vary with weather conditions?


```python
df["weathersit"].value_counts().reindex(
    [1, 2, 3, 4], fill_value=0
)
```

- There were no observations in the severe weather category, so it does not appear in the chart.

```python

# Rename weather categories
weather_labels = {
    1: "Clear / Partly Cloudy",
    2: "Mist / Cloudy",
    3: "Light Rain / Snow",
    4: "Heavy Rain / Snow"
}

df["weather_name"] = df["weathersit"].map(weather_labels)

# Calculate average daily rentals
weather_avg = (
    df.groupby("weather_name", as_index=False)["cnt"]
    .mean()
)

# Keep only weather categories present in the data
weather_order = [
    "Clear / Partly Cloudy",
    "Mist / Cloudy",
    "Light Rain / Snow",
    "Heavy Rain / Snow"
]

weather_order = [
    w for w in weather_order
    if w in weather_avg["weather_name"].values
]

weather_avg["weather_name"] = pd.Categorical(
    weather_avg["weather_name"],
    categories=weather_order,
    ordered=True
)

weather_avg = weather_avg.sort_values("weather_name")

# Weather-based colors
weather_colors = {
    "Clear / Partly Cloudy": "#F4C542",
    "Mist / Cloudy": "#A8B0BC",
    "Light Rain / Snow": "#38A3DB",
    "Heavy Rain / Snow": "#5B3FAD"
}

# Plot
plt.figure(figsize=(11, 5))

ax = sns.barplot(
    data=weather_avg,
    x="weather_name",
    y="cnt",
    hue="weather_name",
    order=weather_order,
    hue_order=weather_order,
    palette=weather_colors,
    errorbar=None,
    legend=False
)

for container in ax.containers:
    ax.bar_label(container, fmt="%.0f", padding=4)

plt.title("Average Daily Bike Rentals by Weather", fontsize=14)
plt.xlabel("Weather Condition")
plt.ylabel("Average Daily Rentals")
plt.xticks(rotation=0)
plt.margins(y=0.12)

sns.despine()
plt.tight_layout()
plt.show()
```

- Average daily rentals were highest in clear or partly cloudy weather (4,877), followed by mist or cloudy conditions (4,036). Light rain or snow had the lowest average, at around 1,803 rentals per day.

## 8. Casual vs. Registered Riders

- Let's compare casual and registered bike rentals on working and non-working days.
- **Business question:** Do casual and registered riders use the bike-sharing service differently depending on the day type?

```python
# Compare the two user groups by day type
user_type_by_day = (
    df.assign(
        day_type=df["workingday"].map({
            1: "Working day",
            0: "Non-working day"
        })
    )
    .groupby("day_type")[["casual", "registered"]]
    .mean()
    .reset_index()
    .melt(
        id_vars="day_type",
        var_name="user_type",
        value_name="avg_rentals"
    )
)

user_type_by_day["user_type"] = user_type_by_day["user_type"].map({
    "casual": "Casual riders",
    "registered": "Registered riders"
})

plt.figure(figsize=(9, 5))

ax = sns.barplot(
    data=user_type_by_day,
    x="day_type",
    y="avg_rentals",
    hue="user_type",
    order=["Working day", "Non-working day"],
    hue_order=["Casual riders", "Registered riders"],
    palette={
        "Casual riders": "#FF6B6B",
        "Registered riders": "#7169E0"
    },
    errorbar=None
)

for container in ax.containers:
    ax.bar_label(container, fmt="%.0f", padding=3)

plt.title("Average Daily Rentals by User Type and Day Type")
plt.xlabel("Day Type")
plt.ylabel("Average Daily Rentals")
plt.legend(title="User Type")

sns.despine()
plt.tight_layout()
plt.show()
```

- Registered riders averaged about 3,978 rentals on working days, compared with 2,959 on non-working days.
- Casual riders averaged 607 rentals on working days and 1,371 on non-working days.
- Registered riders used the service more on working days, while casual riders rented more bikes on non-working days.
- These differences may help the business plan its services and marketing for different customer groups. However, the chart does not explain why customers use the service.

## 9. Temperature and Daily Bike Rentals

- Let's see how daily bike rentals change with temperature.
- **Business question:** Do warmer days tend to have higher bike rental demand?

```python

# Calculate the correlation
temp_corr = df["temp"].corr(df["cnt"])

print(f"Correlation between temperature and rentals: {temp_corr:.2f}")

# Plot temperature against daily rentals
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df,
    x="temp",
    y="cnt",
    hue="season_name",
    hue_order=["Spring", "Summer", "Fall", "Winter"],
    palette=season_colors,
    alpha=0.7,
    s=45
)

# Add a trend line
sns.regplot(
    data=df,
    x="temp",
    y="cnt",
    scatter=False,
    ci=None,
    color="#343A40",
    line_kws={"linewidth": 2, "linestyle": "--"}
)

plt.title("Daily Bike Rentals vs. Temperature")
plt.xlabel("Temperature (normalized)")
plt.ylabel("Total Daily Rentals")
plt.legend(title="Season")

sns.despine()
plt.tight_layout()
plt.show()
```

- Temperature and daily bike rentals have a positive correlation of 0.63.
- In general, warmer days tend to have more bike rentals.
- However, rental demand varies even at similar temperatures.
- Other factors, such as season, weather, and working days, may also affect demand.

## 10. Correlation Between Weather Factors and Bike Rentals

- Let's compare temperature, humidity, wind speed, and daily bike rentals to see how these variables are related.
- **Business question:** Which weather-related variables have the strongest relationship with daily bike rental demand?

```python
# Select numerical variables for the analysis
corr_columns = [
    "temp",
    "atemp",
    "hum",
    "windspeed",
    "cnt"
]

corr_matrix = df[corr_columns].corr()

# Create a correlation heatmap
plt.figure(figsize=(9, 6))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="RdYlBu_r",
    vmin=-1,
    vmax=1,
    center=0,
    linewidths=1,
    square=True
)

plt.title("Correlation Between Weather Factors and Bike Rentals")
plt.tight_layout()
plt.show()
```

- Temperature has a positive correlation with bike rentals (0.63).
- Humidity (-0.10) and wind speed (-0.23) have weak negative correlations with rentals.
- Temperature and feeling temperature are highly correlated (0.99), meaning they provide very similar information.
- These results show relationships between the variables, but they do not prove cause and effect. Other factors, such as season and working days, may also affect demand.
